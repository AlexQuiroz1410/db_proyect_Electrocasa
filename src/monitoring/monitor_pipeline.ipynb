```python
from pyspark.sql.functions import col

catalog = "dbelectrocasa"

df_quarantine = spark.table(
    f"{catalog}.silver.slv_quarantine"
)

total_rechazados = df_quarantine.count()

print(
    f"Registros actualmente en cuarentena: "
    f"{total_rechazados}"
)

display(
    df_quarantine
    .groupBy(
        "source_table",
        "motivo_rechazo"
    )
    .count()
    .orderBy(
        col("count").desc()
    )
)
```
